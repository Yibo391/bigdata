# V4 Task 2 — fine-tuned ModernBERT

This notebook evaluates a new Task 2 model with five user-disjoint folds. It does not change V1, V2, V3, or create a leaderboard submission. Completed folds are reused if the run is interrupted.

```python
from pathlib import Path
import json
import pandas as pd
import torch
import v4_finetuned_factor as v4

print('PyTorch:', torch.__version__)
print('MPS available:', torch.backends.mps.is_available())
assert Path('train.xlsx').exists(), 'Open this notebook from the bigdata-main folder.'
assert Path('outputs/v2_factor_ensemble/oof_predictions.npz').exists(), 'Run the existing V2 evaluation first.'
```

## Train and evaluate

Run this cell once. On the first run it downloads ModernBERT. Five folds may take roughly 1–2 hours on an Apple-silicon Mac. If VS Code is still showing a moving progress bar, it is working. If the kernel stops, run this cell again; completed folds will be skipped.

```python
cfg = v4.V4Config(
    epochs=5,
    max_length=1024,
    resume_completed_folds=True,
)
metrics = v4.run_v4_oof('train.xlsx', cfg)
metrics
```

## Read the final local result

The important number is `v4_v2_calibrated_macro_f1`. Do not create a submission yet; first compare this value with the V3 local score of about 0.449.

```python
result_path = Path('outputs/v4_finetuned_factor/oof_metrics.json')
result = json.loads(result_path.read_text())
print('V4 alone Macro F1:', round(result['v4_only_fixed_quota_macro_f1'], 4))
print('V4 + V2 Macro F1:', round(result['v4_v2_calibrated_macro_f1'], 4))
print('Average gold labels:', round(result['average_gold_labels'], 3))
print('Average predicted labels:', round(result['average_predicted_labels'], 3))
```
